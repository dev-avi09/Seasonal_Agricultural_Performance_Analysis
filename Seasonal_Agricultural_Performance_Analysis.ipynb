10. How does agricultural performance vary across different seasons?

```python
season_performance = df.groupby('Season')[[
    'Yield_Tonnes_Ha',
    'Production_Tonnes',
    'Revenue_INR',
    'Profit_INR'
]].mean().round(2)

season_performance
```

```python
season_performance.plot(kind='bar', figsize=(10,6))
plt.title('Agricultural Performance Across Seasons')
plt.xlabel('Season')
plt.ylabel('Average Value')
plt.xticks(rotation=0)
plt.legend(title='Performance Measures')
plt.show()
```

11. What major seasonal patterns can be observed?

```python
seasonal_patterns = df.groupby('Season')[[
    'Rainfall_mm',
    'Avg_Temperature_C',
    'Humidity_pct',
    'Sunlight_Hours_Day',
    'Soil_Moisture_pct',
    'Fertilizer_kg_ha',
    'Pesticide_Litre_ha',
    'Water_Used_m3',
    'Disease_Pest_Risk_pct'
]].mean().round(2)

seasonal_patterns
```

```python
plt.figure(figsize=(12, 6))

sns.heatmap(
    seasonal_patterns.T,
    annot=True,
    fmt='.2f',
    cmap='YlGnBu'
)
plt.title('Seasonal Patterns in Environmental Conditions and Resource Usage')
plt.xlabel('Season')
plt.ylabel('Variables')
plt.show()
```

12. Which characteristics change between seasons?

```python
characteristics = [
    'Rainfall_mm',
    'Avg_Temperature_C',
    'Humidity_pct',
    'Sunlight_Hours_Day',
    'Soil_pH',
    'Soil_Moisture_pct',
    'Nitrogen_kg_ha',
    'Phosphorus_kg_ha',
    'Potassium_kg_ha',
    'Fertilizer_kg_ha',
    'Pesticide_Litre_ha',
    'Seed_Quality_Score',
    'Water_Used_m3',
    'Disease_Pest_Risk_pct'
]

seasonal_characteristics = df.groupby('Season')[characteristics].mean().round(2)

seasonal_characteristics
```

```python
percentage_change = (
    (seasonal_characteristics.max() - seasonal_characteristics.min())
    / seasonal_characteristics.min() * 100
).sort_values(ascending=False).round(2)

percentage_change
```

```python
plt.figure(figsize=(10, 7))

percentage_change.sort_values().plot(kind='barh')

plt.title('Percentage Variation in Characteristics Across Seasons')
plt.xlabel('Percentage Variation (%)')
plt.ylabel('Characteristic')

plt.show()
```

16. How do economic outcomes vary across seasons?

```python
economic_outcomes = df.groupby('Season')[[
    'Market_Price_INR_Tonne',
    'Total_Cost_INR',
    'Revenue_INR',
    'Profit_INR'
]].mean().round(2)

economic_outcomes
```

```python
economic_outcomes.plot(
    kind='bar',
    figsize=(11, 6)
)
plt.title('Economic Outcomes Across Seasons')
plt.xlabel('Season')
plt.ylabel('Average Value (INR)')
plt.xticks(rotation=0)
plt.legend(title='Economic Measure')
plt.show()
```

```python
profit_margin = (
    df.groupby('Season')['Profit_INR'].sum()
    / df.groupby('Season')['Revenue_INR'].sum()
    * 100
).round(2)

profit_margin
```

```python
profit_margin.plot(
    kind='bar',
    figsize=(8, 5)
)
plt.title('Profit Margin Across Seasons')
plt.xlabel('Season')
plt.ylabel('Profit Margin (%)')
plt.xticks(rotation=0)
plt.show()
```

## 19. What insights can be derived from the observed seasonal differences?

### Insight 1 — Kharif performs the best overall

- Kharif has the highest average yield, production, revenue and profit.
- Its average profit is **₹178,914.65**, with a **25.17% profit margin**.

### Insight 2 — Zaid faces the most economic pressure

- Zaid has the lowest average yield and revenue.
- It also has the highest average cost at **₹543,976.73**.
- Its average profit is **−₹24,804.82**.

### Insight 3 — Environmental conditions change a lot

- Kharif has the highest rainfall, humidity and soil moisture.
- Zaid has the highest temperature and sunlight, but the lowest rainfall and soil moisture.

### Insight 4 — Water usage changes more than fertilizer usage

- Zaid uses the most water, at **6,419.89 m³** on average.
- Fertilizer usage stays fairly similar across all three seasons.

### Insight 5 — Seasonal performance depends on the region

- Kharif performs best in several states, but not everywhere.
- Punjab performs best in Rabi, while Karnataka performs slightly better in Zaid.

### Insight 6 — Environmental factors alone do not explain yield

- Rainfall, temperature, humidity and soil moisture all show very weak correlations with yield.
- This suggests that yield depends on more than just individual environmental factors.

### Insight 7 — Zaid has the highest share of loss-making farms

- Kharif: **42.21%**
- Rabi: **51.14%**
- Zaid: **64.48%**

## 20. What conclusions can reasonably be drawn from the available data?

The analysis shows clear differences in agricultural performance across Kharif, Rabi and Zaid.

- **Kharif** has the strongest overall performance, with the highest average yield, production, revenue and profit.
- **Rabi** shows moderate performance, with positive average profitability.
- **Zaid** has the lowest average yield and revenue, along with the highest average cost and a negative average profit.
- Environmental conditions also vary considerably between the three seasons.
- Zaid uses the most water on average, while fertilizer usage remains fairly similar across seasons.
- Seasonal performance is not the same across all regions, showing that location also matters.
- The correlation results show very weak relationships between individual environmental factors and yield.

The results suggest that agricultural performance depends on a combination of seasonal, regional, environmental and farming factors rather than one factor alone.

These findings should be treated as observed patterns and relationships in the dataset, rather than proof of direct cause and effect.
